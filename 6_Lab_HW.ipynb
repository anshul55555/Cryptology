One-Time Pad (OTP) — Encryption and Decryption

```python
def otp_encrypt(text, key):
    result = ""

    for i in range(len(text)):
        p = ord(text[i].upper()) - ord('A')
        k = ord(key[i].upper()) - ord('A')

        c = (p + k) % 26
        result += chr(c + ord('A'))

    return result


def otp_decrypt(text, key):
    result = ""

    for i in range(len(text)):
        c = ord(text[i].upper()) - ord('A')
        k = ord(key[i].upper()) - ord('A')

        p = (c - k) % 26
        result += chr(p + ord('A'))

    return result


# Input validation
text = input("Enter plaintext (letters only): ").strip().upper()

while not text or not text.isalpha() or not text.isascii():
    print("Error: Enter non-empty English letters only.")
    text = input("Enter plaintext (letters only): ").strip().upper()

while True:
    key = input("Enter key (same length as plaintext): ").strip().upper()

    if not key.isascii() or not key.isalpha():
        print("Error: Key must contain English letters only.")
    elif len(key) != len(text):
        print("Error: Key length must equal plaintext length.")
    else:
        break

ciphertext = otp_encrypt(text, key)

print("Plaintext:", text)
print("Key:", key)
print("Encrypted text:", ciphertext)
print("Decrypted text:", otp_decrypt(ciphertext, key))
```

Brute-Force Attack on Shift Cipher

```python
def shift_decrypt(text, key):
    result = ""

    for ch in text:
        if ch.isalpha() and ch.isascii():
            base = ord('A') if ch.isupper() else ord('a')
            result += chr((ord(ch) - base - key) % 26 + base)
        else:
            result += ch

    return result


# Given ciphertext
text = "MCI RWR WH"

print("Ciphertext:", text)
print("\nBrute-force attack results:\n")

for key in range(26):
    plaintext = shift_decrypt(text, key)
    print("Key:", key, "->", plaintext)
```

Rail Fence Cipher — Encryption

```python
def rail_fence_encrypt(text, rails):
    if rails < 2:
        raise ValueError("Number of rails must be at least 2.")

    fence = [""] * rails
    row = 0
    direction = 1

    for ch in text:
        fence[row] += ch

        if row == 0:
            direction = 1
        elif row == rails - 1:
            direction = -1

        row += direction

    return "".join(fence)


# Input validation
text = input("Enter plaintext: ")

while text.strip() == "":
    print("Error: Plaintext cannot be empty.")
    text = input("Enter plaintext: ")

while True:
    try:
        rails = int(input("Enter number of rails: "))

        if 2 <= rails <= len(text):
            break

        print("Error: Rails must be at least 2 and no greater than the text length.")

    except ValueError:
        print("Error: Enter a valid integer.")


ciphertext = rail_fence_encrypt(text, rails)

print("Plaintext:", text)
print("Number of rails:", rails)
print("Encrypted text:", ciphertext)
```
